```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def show(org,sobel,prewit,cannny):
  plt.figure(figsize=(12,10))
  plt.subplot(2,2,1)
  plt.imshow(org,cmap="gray")
  plt.title("original grayscale image")
  plt.axis("off")
  plt.subplot(2,2,2)
  plt.imshow(sobel,cmap="gray")
  plt.title("sobel image")
  plt.axis("off")
  plt.subplot(2,2,3)
  plt.imshow(prewit,cmap="gray")
  plt.title("prewit image")
  plt.axis("off")
  plt.subplot(2,2,4)
  plt.imshow(canny,cmap="gray")
  plt.title("canny image")
  plt.axis("off")
  plt.tight_layout()
  plt.savefig("cv4")
  plt.show()
```

```python
img=cv2.imread('/content/drive/MyDrive/cv pics/lab3.png')
"""img=cv2.imread("/content/image_ac564628.png")"""
```

```python
sobelx=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
sobely=cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)
```

```python
smag=cv2.magnitude(sobelx,sobely)
```

```python
kernelx=np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
kernely=np.array([[1,0,-1],[1,0,-1],[1,0,-1]])
px=cv2.filter2D(img,-1,kernelx)
py=cv2.filter2D(img,-1,kernely)
pc=px+py
```

```python
canny=cv2.Canny(img,100,150)
```

```python
show(img,np.uint8(np.absolute(smag)),pc,canny)
```
